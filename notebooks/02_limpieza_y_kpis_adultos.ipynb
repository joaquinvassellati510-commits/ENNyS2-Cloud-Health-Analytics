# Limpieza y KPIs – Población Adulta ENNyS2

Objetivo: filtrar población adulta y construir indicadores nutricionales
(exceso de peso y obesidad) con análisis por región, sexo y nivel educativo.


```python
import pandas as pd
```

# ***Dataset***

```python
df = pd.read_csv("ennys2_encuesta.zip", low_memory=False)
```

## ***Variables*** ***relevantes***

```python
columnas = [
    "IMC",
    "edad_salida",
    "antropo_sex",
    "region",
    "Nivel_educacion_total"
]

df_analitico = df[columnas].copy()

df_analitico.shape
```

IMC



```python
df_analitico["IMC"] = pd.to_numeric(df_analitico["IMC"], errors="coerce")

df_analitico["IMC"].describe()
```

Filtro población adulta



```python
df_adultos = df_analitico[df_analitico["edad_salida"] == " "].copy()

df_adultos.shape
```

Variables nutricionales




```python
df_adultos["exceso_peso"] = (df_adultos["IMC"] >= 25).astype(int)
df_adultos["obesidad"] = (df_adultos["IMC"] >= 30).astype(int)
```

KPI Nacional




```python
exceso_nacional = df_adultos["exceso_peso"].mean()
obesidad_nacional = df_adultos["obesidad"].mean()

exceso_nacional, obesidad_nacional
```

Ranking por región




```python
exceso_region = (
    df_adultos
    .groupby("region")["exceso_peso"]
    .mean()
    .sort_values(ascending=False)
)

exceso_region
```

Exceso de peso por sexo



```python
df_adultos.groupby("antropo_sex")["exceso_peso"].mean()
```

Obesidad por sexo







```python
df_adultos.groupby("antropo_sex")["obesidad"].mean()
```

 Exceso de peso por nivel educativo



```python
df_adultos.groupby("Nivel_educacion_total")["exceso_peso"].mean().sort_values(ascending=False)
```

Obesidad por nivel educativo




```python
df_adultos.groupby("Nivel_educacion_total")["obesidad"].mean().sort_values(ascending=False)
```


## Resultados preliminares – Adultos (18+)

- Exceso de peso nacional: 36,1%
- Obesidad nacional: 17,7%

### Diferencias regionales
- Mayor prevalencia: Patagonia
- Menor prevalencia: GBA

### Diferencias por sexo
- Mayor prevalencia en mujeres

### Diferencias por nivel educativo
- Mayor obesidad en población con menor nivel educativo

## Nota metodológica

Los indicadores fueron calculados sin aplicar ponderadores muestrales.
Por lo tanto, los resultados representan proporciones simples de la muestra
